```python
#Q6 
global_meadian_hp = data['horsepower'].median()
global_mode_hp = data['horsepower'].mode()[0]
data['horsepower'] = data['horsepower'].fillna(global_mode_hp)
post_fill_meadian_hp = data['horsepower'].median()

print(post_fill_meadian_hp - global_meadian_hp)
```

```python
asia_cars = data[data['origin'] == 'Asia'][['vehicle_weight', 'model_year']]
X = asia_cars.iloc[:7].values
XTX = X.T.dot(X)
inv = np.linalg.inv(XTX)
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
w = inv.dot(X.T).dot(y)
print(f"What's the sum of all the elements of the result? {w.sum()}")
```
